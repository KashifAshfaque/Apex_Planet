```python
Q1 = df[["Age", "Quantity", "Unit_Price", "Total_Sales"]].quantile(0.25)

Q3 = df[["Age", "Quantity", "Unit_Price", "Total_Sales"]].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR

upper_bound = Q3 + 1.5 * IQR

outliers = (
    (df[["Age", "Quantity", "Unit_Price", "Total_Sales"]] < lower_bound) |
    (df[["Age", "Quantity", "Unit_Price", "Total_Sales"]] > upper_bound)
)

outliers.sum()
```
