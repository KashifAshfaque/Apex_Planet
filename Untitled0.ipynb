## 2. SQL for Business Questions

SQL is used to answer important business questions from the sales data using filtering, aggregation, grouping, sorting, and joins.

```python
import sqlite3
conn = sqlite3.connect("sales_database.db")
df.to_sql("sales", conn, if_exists="replace", index=False)
```

```python
query = """
SELECT *
FROM sales
LIMIT 5;
"""

pd.read_sql(query, conn)
```

### Business Question 1:
**What is the total sales generated by each product?**

```python
query = """
SELECT Product,
       SUM(Total_Sales) AS Total_Revenue
FROM sales
GROUP BY Product
ORDER BY Total_Revenue DESC
"""

product_sales = pd.read_sql(query, conn)

product_sales
```

### Insight

The query compares total revenue generated by each product. Products are sorted from highest to lowest revenue, which helps identify the products contributing the most to overall sales.

### Business Question 2:
**Which are the top 5 products based on total sales?**


```python
query = """
SELECT Product,
       SUM(Total_Sales) AS Total_Revenue
FROM sales
GROUP BY Product
ORDER BY Total_Revenue DESC
LIMIT 5
"""

top_products = pd.read_sql(query, conn)

top_products
```

### Business Question 3:
**Which cities generate the highest total sales?**

```python
query = """
SELECT City,
       SUM(Total_Sales) AS Total_Revenue
FROM sales
GROUP BY City
ORDER BY Total_Revenue DESC
"""

city_sales = pd.read_sql(query, conn)

city_sales
```


### Business Question 4:
**How do total sales change month by month?**

```python
query = """
SELECT
    strftime('%Y-%m', Order_Date) AS Month,
    SUM(Total_Sales) AS Total_Revenue
FROM sales
GROUP BY Month
ORDER BY Month
"""

monthly_sales = pd.read_sql(query, conn)

monthly_sales
```

```python
plt.figure(figsize=(10, 5))

plt.plot(monthly_sales["Month"],
         monthly_sales["Total_Revenue"],
         marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Revenue")

plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

### Business Question 5:
**Which product categories generate the highest total sales?**

```python
query = """
SELECT Category,
       SUM(Total_Sales) AS Total_Revenue
FROM sales
GROUP BY Category
ORDER BY Total_Revenue DESC
"""

category_sales = pd.read_sql(query, conn)

category_sales
```

```python
plt.figure(figsize=(8, 5))

plt.bar(category_sales["Category"],
         category_sales["Total_Revenue"],
         edgecolor="black")

plt.title("Revenue by Product Category")
plt.xlabel("Category")
plt.ylabel("Total Revenue")

plt.xticks(rotation=45)
plt.show()
```

### Business Question 6:
**What is the average order value for each product category?**

```python
query = """
SELECT Category,
       AVG(Total_Sales) AS Average_Order_Value
FROM sales
GROUP BY Category
ORDER BY Average_Order_Value DESC
"""

average_order = pd.read_sql(query, conn)

average_order
```

### Business Question 7:
**How does total sales differ between male and female customers?**

```python
query = """
SELECT Gender,
       SUM(Total_Sales) AS Total_Revenue
FROM sales
GROUP BY Gender
ORDER BY Total_Revenue DESC
"""

gender_sales = pd.read_sql(query, conn)

gender_sales
```

```python
plt.figure(figsize=(7, 4))

plt.bar(gender_sales["Gender"],
        gender_sales["Total_Revenue"],
        edgecolor="black")

plt.title("Revenue by Gender")
plt.xlabel("Gender")
plt.ylabel("Total Revenue")

plt.show()
```
